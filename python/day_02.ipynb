```python
customer_adding = True
shopping_list = []

while customer_adding:
    customer_out = input("Do you want to add or remove? \n Press 'A' for adding \n 'R' for removing \n 'S' to stop creating shopping list")

    if customer_out.lower()=="a":
        print(f"shopping list {shopping_list}")
        item = input("What do you want to add?")
        shopping_list.append(item)

    elif customer_out.lower()=="r":
        print(f"shopping list {shopping_list}")
        item = input("What do you want to remove?")
        if item in shopping_list:
            shopping_list.remove(item)
        else:
            print(f"{item} is not present")

    elif customer_out.lower()=="s":
        customer_adding = False

    else:
        print("Invalid input")
shopping_list.sort()

print(f"shopping list {shopping_list}")
```
